```python
%sql
CREATE OR REPLACE TABLE silver.country (

    country_sk BIGINT NOT NULL GENERATED BY DEFAULT AS IDENTITY comment 'Primary Key',
    country_id BIGINT NOT NULL COMMENT 'Country ID',
    country_name STRING NOT NULL COMMENT 'Country Name',

    etl_filename STRING NOT NULL COMMENT 'etl_filename',
    etl_business_key STRING NOT NULL COMMENT 'business_key',
    etl_record_hash STRING NOT NULL COMMENT 'hash column',
    etl_load_timestamp TIMESTAMP NOT NULL DEFAULT NOW() COMMENT 'Load Time',
    etl_update_timestamp TIMESTAMP NOT NULL DEFAULT NOW() COMMENT 'Updated Time',

CONSTRAINT pk_country_sk PRIMARY KEY (country_sk))
USING DELTA
TBLPROPERTIES('delta.feature.allowColumnDefaults' = 'supported')
COMMENT 'Country Table'
```
