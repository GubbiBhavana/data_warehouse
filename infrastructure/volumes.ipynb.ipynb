```python
%sql
USE CATALOG industry;
```

```python
volumes = ["customer", "location", "products", "sales", "checkpoints"]
```

# Delete the Volumes

```python
for volume in volumes:
    spark.sql(f"DROP VOLUME IF EXISTS landing.{volume}")
    print(f"{volume} deleted successfully")
```

# Create the Volumes

```python
for volume in volumes:
    spark.sql(f"CREATE VOLUME landing.{volume} COMMENT '{volume} data'")
    print(f"{volume} created successfully")
```
