```python
%sql
USE CATALOG industry;
```

```python
schemas = ["bronze", "silver", "gold", "landing"]
```

# Delete the Schemas

```python
for schema in schemas:
    spark.sql(f"DROP SCHEMA IF EXISTS {schema} CASCADE")
    print(f"{schema} deleted successfully")
```

# Create the Schemas

```python
for schema in schemas:
    spark.sql(f"CREATE SCHEMA {schema} COMMENT '{schema} data'")
    print(f"{schema} created successfully")
```
