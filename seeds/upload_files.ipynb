```python
from os import getcwd, walk, path
```

```python
seeds_directory = getcwd()
```

# Upload Data to respective volumes

```python
file_paths = []
for root, _, files in walk(seeds_directory):
    for file in files:
        if file.endswith(".csv"):
            file_paths.append(path.relpath(root) + "/" + file)
file_paths
```

```python
common_path = "/Volumes/industry/landing/"
for file in file_paths:
    print("Working on the file: ", file)
    with open(file ,"r", newline="") as csv_file:
        csv_content = csv_file.read()
        full_path = common_path + file
        dbutils.fs.put(full_path, csv_content, overwrite= True)
    
```
