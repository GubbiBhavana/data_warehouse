```python
from pyspark.sql.functions import col, now
```

```python
%sql
USE CATALOG industry;
```

# Read the Stream

```python
schema = "country_no STRING, country_name STRING, etl_filename STRING, etl_load_timestamp TIMESTAMP, etl_update_timestamp TIMESTAMP"

df = (
    spark.read.format("csv")
    .option("header", True)
    .schema(schema)
    .load("/Volumes/industry/landing/location/*.csv")
    .withColumn("etl_filename",col("_metadata.file_path"))
    .withColumn("etl_load_timestamp", now())
    .withColumn("etl_update_timestamp", now())
)
```

# Write the Stream

```python
(df.write
 .format("delta")
 .saveAsTable("bronze.country"))
```
