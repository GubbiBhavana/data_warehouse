```python
from pyspark.sql.functions import col
```

```python
%sql
USE CATALOG industry;
```

# Read the Stream

```python
schema = "country_id STRING, country_name STRING, etl_filename STRING"

df = (
    spark.readStream.format("cloudFiles")
    .option("cloudFiles.format", "csv")
    .option("header", True)
    .schema(schema)
    .load("/Volumes/industry/landing/location/erp/*.csv")
    .withColumn("etl_filename",col("_metadata.file_path"))
)
```

# Write the Stream

```python
(df.writeStream
 .format("delta")
 .option("checkpointLocation", "/Volumes/industry/landing/checkpoints/location_erp")
 .trigger(availableNow=True)
 .outputMode("append")
 .toTable("bronze.location"))
```
